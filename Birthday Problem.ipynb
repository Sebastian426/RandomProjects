# Imports

```python
import numpy as np
import numpy.random as rnd
import matplotlib.pyplot as plt
```

# Birthday Paradox Function

```python
def birthday_paradox_function(people: int)-> float:
    birthdays = []
    for person in range(people):
        birthdays.append(rnd.randint(0, 366))
    return len(birthdays) != len(set(birthdays))
```

# Simulation

```python
n_sims = 100000
people_array = [10,15,20,25,30,35,40,45,50,55,60]


match_prob = []


for people in people_array:
    matches_count = 0
    for i in range(n_sims):
        matches = birthday_paradox_function(people)
        matches_count += matches
    match_prob.append(matches_count / n_sims)



plt.figure(1)
plt.scatter(people_array, match_prob)
plt.xlabel("Number of People")
plt.ylabel("Match Probability")
plt.show()
```
