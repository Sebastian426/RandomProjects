# Imports

```python
import numpy as np
import numpy.random as rnd
import matplotlib.pyplot as plt
```

# Constants

```python
number_of_runs = 1000000
```

# Monty Hall Sim

```python
stay_wins = 0
switch_wins = 0

for run in range(number_of_runs):
    prize = rnd.randint(1,4)
    pick = rnd.randint(1,4)
    if prize == pick:
        stay_wins += 1
    else:
        switch_wins += 1

print("Stay Wins:", stay_wins)
print("Switch Wins:", switch_wins)

print("Stay probability:", stay_wins / number_of_runs)
print("Switch probability:", switch_wins / number_of_runs)
```

# Monty Hall Function

```python
def monty_hall_function(runs: int) -> tuple:
    stay, switch = 0, 0

    for run in range(runs):
        prize = rnd.randint(1,4)
        pick = rnd.randint(1,4)

        if prize == pick:
            stay += 1
        else:
            switch += 1

    stay_prob, switch_prob = stay/runs, switch/runs
    return stay_prob, switch_prob
```

```python
stay_probabilities, switch_probabilities = [], []

run_list = [10,100,1000,10000,100000,1000000]

for run in run_list:
    stay_prob, switch_prob = monty_hall_function(run)

    stay_probabilities.append(stay_prob)
    switch_probabilities.append(switch_prob)
```

```python
plt.figure(1)
plt.scatter(run_list, stay_probabilities, color = 'red', label = 'stay')
plt.scatter(run_list, switch_probabilities, color = 'blue', label = 'switch')

plt.legend(loc = "upper right", edgecolor = "none")
plt.minorticks_on()
plt.tick_params(top = True,bottom = True,left = True,right = True, which = "minor", direction = "in")
plt.tick_params(top = True,bottom = True,left = True,right = True, which = "major", direction = "in")
plt.hlines(1/3, run_list[0], run_list[-1], color = 'red', label = 'stay', alpha = 0.5)
plt.hlines(2/3, run_list[0], run_list[-1], color = 'blue', label = 'switch', alpha = 0.5)

print(f"for the {run_list[-1]} run trial, the probability of winning if you switch is {switch_probabilities[-1]:.5f}")
plt.show()
```
