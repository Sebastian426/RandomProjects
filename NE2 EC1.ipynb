# Imports

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Functions

```python
def LLEGC(time, investment, waste_cost, operations_maintenance, fuel_cost, decommissioning, energy_generated, discount_rate):

    numerator = 0
    denominator = 0

    for t in range(len(time)):
        numerator += (investment[t]+operations_maintenance[t]+fuel_cost[t]+waste_cost[t]+decommissioning[t])*((1+discount_rate)**(-time[t]))
        denominator += energy_generated[t]*((1+discount_rate)**(-time[t]))

    lifetime_levelised_electricity_generating_cost = numerator/denominator

    return lifetime_levelised_electricity_generating_cost
```

# Parameters

```python
length_of_year = 8736

coal_construction_time = 4
gas_construction_time = 3
nuclear_construction_time = 5

coal_total_years = coal_construction_time + 40
gas_total_years = gas_construction_time + 40
nuclear_total_years = nuclear_construction_time + 40

coal_time = np.arange(1, coal_total_years + 1)
gas_time = np.arange(1, gas_total_years + 1)
nuclear_time = np.arange(1, nuclear_total_years + 1)

##########
#coal data
##########
coal_nominal_capacity = 750
coal_thermal_efficiency = 0.4
coal_station_avalibility = 0.81

coal_investment = np.zeros(coal_total_years, dtype = float)
coal_investment[:coal_construction_time] = 825e6/coal_construction_time

coal_energy_generated = np.zeros(coal_total_years, dtype = float)
coal_energy_generated[coal_construction_time:] = coal_nominal_capacity * length_of_year * coal_station_avalibility

coal_operations_maintenance = np.zeros(coal_total_years, dtype = float)
coal_operations_maintenance[coal_construction_time:] = 6.8 * coal_energy_generated[coal_construction_time:]

coal_fuel_cost = np.zeros(coal_total_years, dtype = float)
coal_fuel_cost[coal_construction_time:] = 6 * coal_energy_generated[coal_construction_time:] * 1/coal_thermal_efficiency

coal_waste_cost = np.zeros(coal_total_years, dtype = float)
coal_waste_cost[coal_construction_time:] = 0

coal_decommissioning_cost = np.zeros(coal_total_years, dtype = float)
coal_decommissioning_cost[-1] = 0



##########
# gas data
##########

gas_nominal_capacity = 600
gas_thermal_efficiency = 0.55
gas_station_avalibility = 0.65

gas_investment = np.zeros(gas_total_years, dtype = float)
gas_investment[:gas_construction_time] = 300e6/gas_construction_time

gas_energy_generated = np.zeros(gas_total_years, dtype = float)
gas_energy_generated[gas_construction_time:] = gas_nominal_capacity * length_of_year * gas_station_avalibility

gas_operations_maintenance = np.zeros(gas_total_years, dtype = float)
gas_operations_maintenance[gas_construction_time:] = 3.8 * gas_energy_generated[gas_construction_time:]

gas_fuel_cost = np.zeros(gas_total_years, dtype = float)
gas_fuel_cost[gas_construction_time:] = 15 * gas_energy_generated[gas_construction_time:] * 1/gas_thermal_efficiency

gas_waste_cost = np.zeros(gas_total_years, dtype = float)
gas_waste_cost[gas_construction_time:] = 0

gas_decommissioning_cost = np.zeros(gas_total_years, dtype = float)
gas_decommissioning_cost[-1] = 0




##############
# nuclear data
##############

nuclear_nominal_capacity = 1000
nuclear_thermal_efficiency = 0.33
nuclear_station_avalibility = 0.86

nuclear_investment = np.zeros(nuclear_total_years, dtype = float)
nuclear_investment[:nuclear_construction_time] = 1500e6/nuclear_construction_time

nuclear_energy_generated = np.zeros(nuclear_total_years, dtype = float)
nuclear_energy_generated[nuclear_construction_time:] = nuclear_nominal_capacity * length_of_year * nuclear_station_avalibility

nuclear_operations_maintenance = np.zeros(nuclear_total_years, dtype = float)
nuclear_operations_maintenance[nuclear_construction_time:] = 8 * nuclear_energy_generated[nuclear_construction_time:]

nuclear_fuel_cost = np.zeros(nuclear_total_years, dtype = float)
nuclear_fuel_cost[nuclear_construction_time:] = 1 * nuclear_energy_generated[nuclear_construction_time:] * 1/nuclear_thermal_efficiency

nuclear_waste_cost = np.zeros(nuclear_total_years, dtype = float)
nuclear_waste_cost[nuclear_construction_time:] = 3 * nuclear_energy_generated[nuclear_construction_time:]

nuclear_decommissioning_cost = np.zeros(nuclear_total_years, dtype = float)
nuclear_decommissioning_cost[-1] = 600e6

# print(nuclear_investment, nuclear_operations_maintenance, nuclear_fuel_cost, nuclear_waste_cost, nuclear_energy_generated)
```

# Calculations

```python
##################################
# 6% interest rate, 40 year period
##################################

coal_LLEGC_ten_six = LLEGC(time = coal_time,
                           investment = coal_investment,
                           waste_cost = coal_waste_cost,
                           operations_maintenance = coal_operations_maintenance,
                           fuel_cost = coal_fuel_cost,
                           energy_generated = coal_energy_generated,
                           decommissioning= coal_decommissioning_cost,
                           discount_rate = 0.06)

gas_LLEGC_ten_six = LLEGC(time = gas_time,
                           investment = gas_investment,
                           waste_cost = gas_waste_cost,
                           operations_maintenance = gas_operations_maintenance,
                           fuel_cost = gas_fuel_cost,
                           energy_generated = gas_energy_generated,
                           decommissioning= gas_decommissioning_cost,
                           discount_rate = 0.06)

nuclear_LLEGC_ten_six = LLEGC(time = nuclear_time,
                           investment = nuclear_investment,
                           waste_cost = nuclear_waste_cost,
                           operations_maintenance = nuclear_operations_maintenance,
                           fuel_cost = nuclear_fuel_cost,
                           energy_generated = nuclear_energy_generated,
                           decommissioning= nuclear_decommissioning_cost,
                           discount_rate = 0.06)

print(coal_LLEGC_ten_six, gas_LLEGC_ten_six, nuclear_LLEGC_ten_six)
```
